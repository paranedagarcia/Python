La **abstracción** es uno de los pilares de la Programación Orientada a Objetos (POO). Consiste en **ocultar los detalles complejos de implementación y mostrar únicamente la interfaz esencial** que el usuario necesita para interactuar con un objeto.

En Python, la forma estándar de aplicar abstracción es mediante el módulo **`abc`** (*Abstract Base Classes*), utilizando la clase base `ABC` y el decorador `@abstractmethod`.

A continuación, tienes **3 ejemplos sencillos e intuitivos** para principiantes, cada uno con su enunciado y solución paso a paso.


### Ejemplo 1: Sistema de Notificaciones

#### **Enunciado del problema:**
Una aplicación necesita enviar alertas a los usuarios a través de distintos canales (por ejemplo, **Email** y **SMS**). La aplicación principal no debe preocuparse por cómo funciona cada red de comunicación; solo necesita un método unificado llamado **`enviar(mensaje)`** que funcione para cualquier tipo de notificación.

* **¿Dónde está la abstracción?** El bucle final no sabe si está enviando un correo o un SMS; solo sabe que cualquier objeto derivado de `Notificacion` responde al método `.enviar()`.

#### **Código en Python:**


```python
from abc import ABC, abstractmethod

# 1. Clase Abstracta (El molde / contrato)
class Notificacion(ABC):

    @abstractmethod
    def enviar(self, mensaje: str) -> None:
        """Método abstracto: obliga a las subclases a definir cómo enviar el mensaje."""
        pass

# 2. Subclases Concretas
class NotificacionEmail(Notificacion):
    def __init__(self, correo_destino: str):
        self.correo_destino = correo_destino

    def enviar(self, mensaje: str) -> None:
        # Oculta la complejidad de la conexión SMTP
        print(f"📧 Enviando Correo a [{self.correo_destino}]: {mensaje}")

class NotificacionSMS(Notificacion):
    def __init__(self, numero_telefono: str):
        self.numero_telefono = numero_telefono

    def enviar(self, mensaje: str) -> None:
        # Oculta la complejidad de la red celular
        print(f"📱 Enviando SMS al [{self.numero_telefono}]: {mensaje}")

# --- Uso del código ---
canales = [
    NotificacionEmail("usuario@ejemplo.com"),
    NotificacionSMS("+56912345678")
]

for canal in canales:
    canal.enviar("Tu código de verificación es 8492")
```

### Ejemplo 2: Procesador de Pagos

#### **Enunciado del problema:**
Una tienda en línea acepta cobros mediante **Tarjeta de Crédito** y **PayPal**. Debes diseñar una estructura donde el carrito de compras solo llame a **`procesar_pago(monto)`** sin importar las pasarelas bancarias o APIs internas que requiera cada medio de pago.

* **¿Dónde está la abstracción?** Se oculta la validación de tarjetas o la autenticación por token en PayPal tras una llamada común `procesar_pago()`.

#### **Código en Python:**



```python
from abc import ABC, abstractmethod

# 1. Clase Abstracta
class MedioPago(ABC):

    @abstractmethod
    def procesar_pago(self, monto: float) -> None:
        """Define el contrato de cobro para cualquier medio de pago."""
        pass

# 2. Subclases Concretas
class TarjetaCredito(MedioPago):
    def __init__(self, numero_tarjeta: str):
        self.numero_tarjeta = numero_tarjeta

    def procesar_pago(self, monto: float) -> None:
        # Muestra solo los últimos 4 dígitos por seguridad
        ultimos_digitos = self.numero_tarjeta[-4:]
        print(f"💳 Procesando cobro bancario de ${monto:.2f} a la tarjeta ****{ultimos_digitos}")

class Paypal(MedioPago):
    def __init__(self, cuenta_email: str):
        self.cuenta_email = cuenta_email

    def procesar_pago(self, monto: float) -> None:
        print(f"🅿️ Autenticando en PayPal ({self.cuenta_email}) y transfiriendo ${monto:.2f}")

# --- Uso del código ---
metodos_pago = [
    TarjetaCredito("4532890123456789"),
    Paypal("cliente@email.com")
]

monto_compra = 150.00

for medio in metodos_pago:
    medio.procesar_pago(monto_compra)
```

### Ejemplo 3: Encendido de Vehículos

#### **Enunciado del problema:**
Queremos simular el arranque de diferentes medios de transporte (**Auto** y **Moto**). Aunque la mecánica interna para arrancar un motor de auto (inyección electrónica) es muy distinta a la de una moto (pedal o encendido manual), el conductor solo requiere la acción unificada de **`arrancar()`**.

* **¿Dónde está la abstracción?** Intentar instanciar directamente la clase base (`v = Vehiculo("Generico")`) lanzará un error en Python (`TypeError`), garantizando que nadie pueda crear un "vehículo abstracto" sin definir cómo arranca.

#### **Código en Python:**

```python
from abc import ABC, abstractmethod

# 1. Clase Abstracta
class Vehiculo(ABC):
    def __init__(self, marca: str):
        self.marca = marca

    @abstractmethod
    def arrancar(self) -> None:
        """Interfaz común para el arranque."""
        pass

# 2. Subclases Concretas
class Auto(Vehiculo):
    def arrancar(self) -> None:
        print(f"🚗 {self.marca}: Inyectando combustible y girando la llave de encendido.")

class Moto(Vehiculo):
    def arrancar(self) -> None:
        print(f"🛵 {self.marca}: Presionando el pedal de arranque e ignición.")

# --- Uso del código ---
garaje = [
    Auto("Toyota"),
    Moto("Honda")
]

for vehiculo in garaje:
    vehiculo.arrancar()
```
