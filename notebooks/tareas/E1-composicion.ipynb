### 1. Sistema de Gestión de Vehículos Autónomos

Diseñar la arquitectura de un VehiculoAutonomo compuesto por un módulo de navegación (SistemaGPS), un conjunto de sensores de entorno (LiDAR, CamaraVision, SensorUltrasonico), un controlador de propulsión (UnidadMotor) y un motor de toma de decisiones (IAConduccion), permitiendo evaluar la ruta y reaccionar ante obstáculos combinando sus componentes.

```python
#Creando clases para los componentes del vehiculo individualmente
class SistemaGPS:
    def __init__(self, ubicacion = "actual"): #Le otorga una ubicación predeterminada al gps
        self.ubicacion = ubicacion
    def coordenadas(self): 
        return f" La ubicacion actual del vehiculo es {self.ubicacion}"

class Lidar:
    def escaneo(self):
        return f"Lidar: Inicializando escaneo. Escaneo finalizado ¡No se encontraron obstaculos!"

class CamaraVision:
    def visualizacion(self):
        return f"CamaraVision: No se encuentran peatones ni ciclitas en la via es seguro"

class SensorUltrasonico:
    def medir_distancia(self):
        return f"SensorUltrasonico: Distancia segura! puede avanzar "

class Unidadmotor:
    def __init__(self):
        self.velocidad = 0 #Incia en 0 porque el vehiculosse encuentra detenido antes de encenderlo
    def aceleracion(self, velocidad): #Asignandole un nuevo valor a la velocidad
        self.velocidad = velocidad 
        return f"Unidadmotor: La velocididad del vehiculo es {self.velocidad} km/h"
    def frenado(self):
        self.velocidad = 0 #La velocidad se mantiene en 0 porque indica que el vehiculo se encuentra detenido
        return f"Unidadmotor: El vehiculo se encuentra detenido"

#Creando clase IACpnduccion
class IAConduccion:
    def __init__(self, gps, lidar, camara, sensor, motor): #instanciando los atributos de sensores y motor
        self.gps = gps
        self.lidar = lidar
        self.camara = camara
        self.sensor = sensor 
        self.motor = motor
    
    def evaluando_y_conduciendo(self):
        print(f"-----IA CONDUCCION-----")
        #Mostrando en pantalla los resultados de cada sensor 
        print(self.gps.coordenadas)
        print(self.lidar.escaneo)
        print(self.camara.visualizacion)
        print(self.sensor.medir_distancia)
        #Mostrando en pantalla que la ruta es segura entonces puede acelerar
        print("Decision: Ruta libre y segura. En marcha!")
        return self.motor.aceleracion(80)

#Creando clase Vehiculoautonomo
class Vehiculoautonomo:
    def __init__(self):
        #Indicicando los componentes que tiene el vehiculo
        self.gps = SistemaGPS()
        self.lidar = Lidar()
        self.camara = CamaraVision()
        self.sensor = SensorUltrasonico()
        self.motor = Unidadmotor()
        self.ia = IAConduccion(self.gps, self.lidar, self.camara, self.sensor, self.motor)
    
    def iniciando_viaje(self):
        print("Inicializando sistema de vehiculo autonomo")
        resultado =self.ia.evaluando_y_conduciendo()
        print(resultado)

#Bucle de control:aisla la prueba del vehiculo autonomo para evitar ejecuciones automaticas sin importar
if __name__ == "__main__":
    mi_vehiculo = Vehiculoautonomo() #instanciando el vehiculo autonomo
    mi_vehiculo.iniciando_viaje() 



"""Maykol Coronado
Hector Tobar
Ignacio Saez
Diana Dusseaux"""
```
