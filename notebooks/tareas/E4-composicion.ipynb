### 4. Sistema de Estación Meteorológica e Instrumental

Diseñar la clase EstacionMeteorologica compuesta por un conjunto independiente de instrumentos de medición (Barometro, Anemometro, Pluviometro, Termohigrometro) y un módulo de transmisión de datos (ModuloTelemetria), permitiendo procesar y enviar reportes climáticos unificados al agregar las lecturas de cada uno de sus sensores.

```python
import random
class Dia:
    def __init__(self):
        self.presion = round(random.uniform(990,1030),2)
        self.viento = round(random.uniform(0,100),2)
        self.precip = round(random.uniform(0,100),2)
        self.temp = round(random.uniform(0,40),2)
        self.humedad = round(random.uniform(0,100),2)

class Barometro:
    def lectura(self,dia):
         print(f"Presion atmosferica: {dia.presion} u")

class Anemometro:
    def lectura(self,dia):
        print(f"Velocidad del viento: {dia.viento} km/h")

class Pluviometro:
    def lectura(self,dia):
        print(f"Precipitacion: {dia.precip} mm")

class Termohigrometro:
    def lectura(self,dia):
        print(f"Temperatura: {dia.temp} C°\n"
                f"Humedad: {dia.humedad} %")

class ModuloTelemetria:
    def __init__(self,barometro,anemometro,pluviometro,termohigrometro):
        self.barometro=barometro
        self.anemometro=anemometro
        self.pluviometro=pluviometro
        self.termohigrometro=termohigrometro


    def Informe(self, dia):
        self.barometro.lectura(dia)
        self.anemometro.lectura(dia)
        self.pluviometro.lectura(dia)
        self.termohigrometro.lectura(dia)

class EstacionMeteorologica:
    def __init__(self,nombre,modulo_telemetria):
        self.nombre = nombre
        self.modulo_telemetria = modulo_telemetria

    def enviar_informe(self,dia):
      print("Analizando medidas Meteorologicas...")
      print("Recopilando Datos...")
      print(f"---Reporte Meteorologico de la estacion {self.nombre} ---")
      self.modulo_telemetria.Informe(dia)
      print("Reporte enviado con exito.")

dia1 = Dia()
dia2 = Dia()
dia3 = Dia()

bar = Barometro()
ane = Anemometro()
plu = Pluviometro()
ter = Termohigrometro()

tel= ModuloTelemetria(bar,ane,plu,ter)

estacion1= EstacionMeteorologica("Castrocolis",tel)

estacion1.enviar_informe(dia1)
estacion1.enviar_informe(dia2)
estacion1.enviar_informe(dia3)
```
