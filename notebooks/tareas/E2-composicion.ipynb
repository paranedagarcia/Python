### 2. Sistema de Gestión Hospitalaria y Fichas Médicas

Implementar una clase FichaMedica constituida por la información personal del paciente (DatosPaciente), el historial de atenciones (ListaConsultas), los registros de constantes vitales (SignosVitales) y un módulo de prescripción (RecetaMedica), facilitando la actualización de diagnósticos y tratamientos de forma modular.

```python
class DatosPaciente:
    def __init__(self, nombre: str, rut: str, edad: int, genero: str):
        self.nombre = nombre
        self.rut = rut
        self.edad = edad
        self.genero = genero

    def __str__(self):
        return (f"Paciente: {self.nombre} | RUT: {self.rut} | Edad: {self.edad} años | Género: {self.genero}")

class SignosVitales:
    def __init__(self, presion_arterial: str = "N/A", frecuencia_cardiaca: int = 0, temperatura: float = 0.0):
        self.presion_arterial = presion_arterial
        self.frecuencia_cardiaca = frecuencia_cardiaca
        self.temperatura = temperatura

    def actualizar(self, presion: str, frecuencia: int, temp: float):
        self.presion_arterial = presion
        self.frecuencia_cardiaca = frecuencia
        self.temperatura = temp

    def __str__(self):
        return(f"Signos Vitales -> Preión: {self.presion_arterial} mmHg | "
                f"FC: {self.frecuencia_cardiaca} bpm | Temperatura: {self.temperatura} °C")

class RecetaMedica:
    def __init__(self):
        self.medicamentos = []

    def agregar_medicamento(self, medicamento: str, dosis: str, indicacion: str):
        self.medicamentos.append({
            "medicamento": medicamento,
            "dosis": dosis,
            "indicacion": indicacion
        })

    def __str__(self):
        if not self.medicamentos:
            return "Receta Médica: Sin medicamentos prescritos."
        
        resultado = "Receta Médica:\n"
        for i, item in enumerate(self.medicamentos, 1):
            resultado += f"  {i}. {item['medicamento']} ({item['dosis']}) - Indicación: {item['indicacion']}\n"
        return resultado.strip()

class ListaConsultas:
    def __init__(self):
        self.consultas = []

    def registrar_atencion(self, fecha: str, medico: str, diagnostico: str):
        self.consultas.append({
            "fecha": fecha,
            "medico": medico,
            "diagnostico": diagnostico
        })

    def __str__(self):
        if not self.consultas:
            return "Historial de Consultas: Sin atenciones registradas."

        resultado = "Historial de Atenciones:\n"
        for i, c in enumerate(self.consultas, 1):
            resultado += f" [{c['fecha']}] Atendido por {c['medico']} -> Diagnostico: {c['diagnostico']}\n"
        return resultado.strip()
```

```python
class FichaMedica:
    def __init__(self, nombre: str, rut: str, edad: int, genero: str):
        self.paciente = DatosPaciente(nombre, rut, edad, genero)
        self.signos_vitales = SignosVitales()
        self.consultas = ListaConsultas()
        self.receta = RecetaMedica()

    def registrar_atencion(self, fecha: str, medico: str, diagnostico: str):
        self.consultas.registrar_atencion(fecha, medico, diagnostico)

    def actualizar_signos(self, presion: str, frecuencia: int, temperatura: float):
        self.signos_vitales.actualizar(presion, frecuencia,temperatura)

    def prescribir_medicamento(self, medicamento: str, dosis: str, indicacion: str):
        self.receta.agregar_medicamento(medicamento, dosis, indicacion)

    def mostrar_ficha_completa(self):
        print("=" * 60)
        print("               FICHA MÉDICA HOSPITALARIA               ")
        print("=" * 60)
        print(f"{self.paciente}\n")
        print(f"{self.signos_vitales}\n")
        print(f"{self.consultas}\n")
        print(f"{self.receta}\n")
        print("=" * 60)
```

```python
if __name__ == "__main__":
    ficha = FichaMedica(
        nombre="Germán Cortés",
        rut="22.446.415-0",
        edad=19,
        genero="Masculino"
    )

    ficha.actualizar_signos(presion="120/80", frecuencia=72, temperatura=36.8)

    ficha.registrar_atencion(
        fecha="23/07/2026", 
        medico="Dr. Rodrigo Bravo", 
        diagnostico="Resfrío común e irritación faríngea"
    )

    ficha.prescribir_medicamento(
        medicamento="Paracetamol", 
        dosis="1g", 
        indicacion="Cada 8 horas por 5 días"
    )
    ficha.prescribir_medicamento(
        medicamento="Ibuprofeno", 
        dosis="4mg", 
        indicacion="Cada 12 horas por 5 días"
    )

    ficha.mostrar_ficha_completa()
```
