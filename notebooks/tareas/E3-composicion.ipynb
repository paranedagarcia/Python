### Motor de Videojuego 2D (Entidad / Personaje de Estrategia)

Crear la clase UnidadEstrategia utilizando composición al integrar un componente de posicionamiento (Transform2D), un componente de renderizado visual (SpriteRenderer), un módulo de físicas y colisiones (Collider2D) y un componente de comportamiento táctico (IACombate), permitiendo armar distintas unidades de juego combinando estos módulos.

```python
# Componente de Posicionamiento
class Transform2D:
    def __init__(self, x, y):
        self.x = x
        self.y = y

# Componente de Renderizado Visual
class SpriteRenderer:
    def __init__(self, imagen):
        self.imagen = imagen

# Componente de Físicas y Colisiones
class Collider2D:
    def __init__(self, ancho, alto):
        self.ancho = ancho
        self.alto = alto

# Componente de Comportamiento Táctico
class IACombate:
    def __init__(self, tipo_ataque):
        self.tipo_ataque = tipo_ataque

    def actuar(self):
        print(f"Ejecutando estrategia de ataque: {self.tipo_ataque}")



class UnidadEstrategia:
    def __init__(self, transform, sprite, collider, ia):
        # La unidad está compuesta por estos 4 módulos
        self.transform = transform
        self.sprite = sprite
        self.collider = collider
        self.ia = ia

    def mover(self, nuevo_x, nuevo_y):
        self.transform.x = nuevo_x
        self.transform.y = nuevo_y
        print(f"Unidad movida a: ({self.transform.x}, {self.transform.y})")

    def ejecutar_turno(self):
        self.ia.actuar()



# Creamos los componentes para un Arquero
posicion_arquero = Transform2D(x=10, y=20)
estetica_arquero = SpriteRenderer("arquero_sprite.png")
fisicas_arquero = Collider2D(ancho=32, alto=32)
ia_arquero = IACombate(tipo_ataque="A distancia")

# juntamos los componentes (Composición)
arquero = UnidadEstrategia(
    transform=posicion_arquero,
    sprite=estetica_arquero,
    collider=fisicas_arquero,
    ia=ia_arquero
)

# Usamos la unidad
arquero.mover(15, 25)
arquero.ejecutar_turno()
```
