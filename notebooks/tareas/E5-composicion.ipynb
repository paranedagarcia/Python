### 5. Sistema de Producción y Ensamblaje Industrial
Implementar una clase LineaEnsamblaje estructurada por una secuencia de estaciones de trabajo (BrazoRobotico, EscanerCalidad, CintaTransportadora) y una unidad de supervisión (ModuloControlSeguridad), logrando coordinar el flujo de manufactura a través de la ejecución sucesiva de sus partes compuestas.

```python
class LineaEnsamblaje:
    def __init__(self):
        """ Inicializa los módulos principales de la línea """
        self.brazo = BrazoRobotico()
        self.escaner = EscanerCalidad()
        self.cinta = CintaTransportadora()
        self.modulo_seguridad = ModuloControlSeguridad()
    def iniciar(self):
        """Simula el inicio del proceso de ensamblaje """
        self.brazo.mover(10) # Mueve el brazo a la posición 10
        self.escaner.escanear(self.cinta.objetos[0]) # Escanea el primer objeto de la cinta 
        self.modulo_seguridad.verificar_seguridad() # Verifica el estado de seguridad


class BrazoRobotico:
    def __init__(self):
        self.posicion = 0  # Posición inicial del brazo

    def mover(self, posicion):
        self.posicion = posicion # Actualiza posición del brazo

class EscanerCalidad:   
    def __init__(self):
        self.resultado = None # Resultado del último escaneo

    def escanear(self, objeto):
        self.resultado = "calidad_buena" if objeto.es_bueno() else "calidad_mala" # Determina si el objeto es malo o bueno

class CintaTransportadora:
    def __init__(self):
        self.objetos = [] # Lista de objetos en la cinta 
    def transportar(self, objeto):
        self.objetos.append(objeto) # Agrega un objeto a la cinta 


class ModuloControlSeguridad:
    def __init__(self):
        self.estado = "seguro" # Estado inicial del sistema
    def verificar_seguridad(self):
        return self.estado == "seguro"  # Devuelve el True si el sistema esta seguro


class Objeto:
    def __init__(self,bueno=True): 
        self.bueno = bueno # Define si el objeto es de buena calidad 
    def es_bueno(self):
        return self.bueno # Retorna el estado de calidad del objeto


"""  Ejecución del sistema  """
linea = LineaEnsamblaje()
objeto1 = Objeto(bueno=True) # Creamos un objeto de buena calidad 
linea.cinta.transportar(objeto1) # Lo transportamos en la cinta 
linea.iniciar() # Iniciamos la línea de ensamblaje 
print (linea.modulo_seguridad.verificar_seguridad()) # Mostramos el estado de seguridad 

```
