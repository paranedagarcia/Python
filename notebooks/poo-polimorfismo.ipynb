# Polimorfismo

### 1. Polimorfismo por Herencia (Sobrescritura de Métodos / Overriding)

En este enfoque, diferentes subclases heredan de una misma clase base y sobrescriben un método común. Una misma función o bucle puede procesar objetos de distintas subclases llamando al mismo método.

```python
class ProcesadorPago:
    def procesar_pago(self, monto: float) -> str:
        raise NotImplementedError("Las subclases deben implementar este método.")

class PagoTarjeta(ProcesadorPago):
    def __init__(self, numero_tarjeta: str):
        self.numero_tarjeta = numero_tarjeta

    def procesar_pago(self, monto: float) -> str:
        return f"💳 Procesando ${monto:.2f} con Tarjeta ending in {self.numero_tarjeta[-4:]}"

class PagoPayPal(ProcesadorPago):
    def __init__(self, email: str):
        self.email = email

    def procesar_pago(self, monto: float) -> str:
        return f"🅿️ Procesando ${monto:.2f} con cuenta PayPal ({self.email})"

class PagoCripto(ProcesadorPago):
    def __init__(self, billetera: str):
        self.billetera = billetera

    def procesar_pago(self, monto: float) -> str:
        return f"🪙 Procesando ${monto:.2f} vía Criptomoneda (Wallet: {self.billetera[:6]}...)"

# --- Demostración de Polimorfismo ---
def ejecutar_cobro(metodo_pago: ProcesadorPago, monto: float):
    # La función no necesita saber qué tipo exacto de pago es, solo invoca el método común
    print(metodo_pago.procesar_pago(monto))

pagos = [
    PagoTarjeta("4532123456788901"),
    PagoPayPal("usuario@email.com"),
    PagoCripto("0x71C7656EC7ab88b098defB751B7401B5f6d8976F")
]

for pago in pagos:
    ejecutar_cobro(pago, 150.0)
```

### 2. Polimorfismo mediante Duck Typing ("Tipado de Pato")
En Python no es obligatorio que las clases compartan una clase madre para ser polimórficas. Siguiendo el principio "si camina como pato y suena como pato, es un pato", basta con que objetos de clases completamente independientes implementen un método con la misma firma.

```python
class ExportadorPDF:
    def exportar(self, datos: dict) -> str:
        return f"📄 Generando documento PDF con los datos: {list(datos.keys())}"

class ExportadorCSV:
    def exportar(self, datos: dict) -> str:
        filas = ", ".join(f"{k}={v}" for k, v in datos.items())
        return f"📊 Generando archivo CSV estructurado: [{filas}]"

class ExportadorJSON:
    def exportar(self, datos: dict) -> str:
        return f"🔣 Generando cadena JSON formateada con {len(datos)} claves."

# --- Demostración de Duck Typing ---
def generar_reporte(exportador, datos: dict):
    # Funciona con cualquier objeto que posea el método 'exportar', sin importar su clase
    print(exportador.exportar(datos))

informe_datos = {"ventas": 1200, "clientes": 45, "periodo": "Q3"}

exportadores = [ExportadorPDF(), ExportadorCSV(), ExportadorJSON()]

for exp in exportadores:
    generar_reporte(exp, informe_datos)
```

### 3. Polimorfismo con Métodos Especiales (Dunder Methods)
Python utiliza métodos especiales (como __add__ o __len__) para permitir que operadores estándar (+, len()) reaccionen de manera polimórfica según la clase del objeto.

```python
class InventarioProducto:
    def __init__(self, nombre: str, stock: int):
        self.nombre = nombre
        self.stock = stock

    # Sobrescritura de suma (+) para combinar stock del mismo producto
    def __add__(self, otro: 'InventarioProducto') -> 'InventarioProducto':
        if isinstance(otro, InventarioProducto) and self.nombre == otro.nombre:
            return InventarioProducto(self.nombre, self.stock + otro.stock)
        raise ValueError("Solo se pueden sumar inventarios del mismo producto.")

    # Representación en cadena polimórfica para print() y str()
    def __str__(self) -> str:
        return f"📦 Producto: {self.nombre} | Stock Disponible: {self.stock} unidades"

# --- Demostración ---
almacen_norte = InventarioProducto("Laptops", 15)
almacen_sur = InventarioProducto("Laptops", 25)

# El operador '+' invoca polimórficamente __add__
total_inventario = almacen_norte + almacen_sur

print(almacen_norte)
print(almacen_sur)
print(f"Resultado tras fusión: {total_inventario}")
```
